```r
# Setup directories
traits.dir = "/nfs/lab/FinnGen_R11/finemap/summary/"
finrich.dir = '/nfs/lab/tscc/welison/FNIH.Liver/11_FINRICH/'
traits.bed.dir = "/nfs/lab/tscc/welison/FNIH.Liver/11_FINRICH//traits/"
```

```r
# Grep all traits
files = dir(traits.dir, pattern = ".SUSIE.snp.filter.tsv")
files = data.frame(files = paste0(traits.dir, files), 
           trait = str_split_fixed(files, ".SUSIE.snp.filter.tsv", n = 2)[,1])
message("all traits found: ", nrow(files))

# Grep traits we need
traits.ls = paste0(finrich.dir, 'FinnGen.trait.list.txt')
traits.ls = fread(traits.ls, header = FALSE)
traits.ls$trait = basename(traits.ls$V1)
traits.ls = traits.ls %>%
  distinct(trait, .keep_all = TRUE)
message("Traits found: ", nrow(traits.ls))
traits.ls$trait = str_split_fixed(traits.ls$trait, '.gz.ldsc.sumstats.gz', n=2)[,1]

# subset files for only traits we need
files = dplyr::filter(files, trait %in% traits.ls$trait)
message("Traits with fine mapped data for: ", nrow(files))
traits.use = files$files
head(files)
```

```r
# Make bed files
for (i in seq_along(traits.use)){
    # Grep files and make names
    file.use = files[i,1]
    trait.use = files[i,2]
    message("Making a bed for ", trait.use)
    out.bed.file = paste(traits.bed.dir, trait.use, "_Hg38.bed", sep = "")
    # Read table
    trait.data = fread(file.use)
    # Manipolate table
    trait.data$variant 
    trait.data$start <- trait.data$position - 1
    trait.data <- trait.data %>%
        dplyr::select(chromosome, start, end=position, v, allele1, allele2, maf, beta, 
                     p, se, trait, region, cs, cs_specific_prob)
    # Write bed
    print(trait.use)
    print(head(trait.data))
    write.table(trait.data,
                file = out.bed.file,
                col.names = F, row.names = F, sep = "\t", quote = F)
}
```

# MVP cALT GWAS

```r
mvp.cred <- read.table('/nfs/lab/projects/nash_nafld_liver/GWAS/MVP_GWAS/NAFLD.EUR.MVP.2021.credset.tsv', 
                      sep='\t', header=T)

dim(mvp.cred)
head(mvp.cred)
```

```r
sum(str_detect(mvp.cred$CS.SNP, 'rs'))
sum(str_detect(mvp.cred$LEAD.SNP, 'rs'))
```

```r
#Select mart
ensembl <- useEnsembl("snp",dataset = "hsapiens_snp") # ,GRCh = "38" automatically connects to hg38

#get genomic position
SNPs <- getBM(attributes=c("refsnp_id",
                   "chr_name",
                   "chrom_start",
                   "chrom_end"),
    filters ="snp_filter", values =mvp.cred$CS.SNP, mart = ensembl, uniqueRows=TRUE)

dim(SNPs)

SNPs.filt <- filter(SNPs, chr_name %in% as.character(1:22))

dim(SNPs.filt)
head(SNPs.filt)
```

```r
mvp.cred.hg38 <- SNPs.filt

mvp.cred.hg38$chrom_start <- mvp.cred.hg38$chrom_end - 1
mvp.cred.hg38 <- dplyr::select(mvp.cred.hg38, CS.SNP=refsnp_id, chr.hg38=chr_name, start.hg38=chrom_start, end.hg38=chrom_end)

mvp.cred.hg38 <- left_join(mvp.cred, mvp.cred.hg38)

dim(mvp.cred.hg38)
head(mvp.cred.hg38)
```

```r
mvp.cred.hg38.bed <- mvp.cred.hg38
mvp.cred.hg38.bed$variant = paste0(mvp.cred.hg38.bed$chr.hg38,":",mvp.cred.hg38.bed$end.hg38,":",
                              mvp.cred.hg38.bed$EA,":",mvp.cred.hg38.bed$NEA)
mvp.cred.hg38.bed$PP <- as.numeric(str_remove(mvp.cred.hg38.bed$PP, "%")) / 100
mvp.cred.hg38.bed$chr.hg38 <- paste0('chr',mvp.cred.hg38.bed$chr.hg38)
mvp.cred.hg38.bed <- dplyr::select(mvp.cred.hg38.bed, chr=chr.hg38, start=start.hg38, end=end.hg38, rsid=CS.SNP, variant, 
              p=P, beta=Beta, cs_id=LEAD.SNP, prob=PP)

dim(mvp.cred.hg38.bed)
head(mvp.cred.hg38.bed)
```

```r
write.table(mvp.cred.hg38,'/nfs/lab/projects/nash_nafld_liver/GWAS/MVP_GWAS/NAFLD.EUR.MVP.2021.credset.hg38.tsv', 
                      sep='\t', col.names=T, row.names=F, quote=F)
```

# DESeq Reuslts Bed

```r
deseq.outdir <- '/nfs/lab/projects/nash_nafld_liver/downstream_all/02_DESEQ_ATAC/'
bed.dir <- '/nfs/lab/tscc/welison/FNIH.Liver/11_FINRICH/sites/'
```

```r
for (file in list.files(deseq.outdir, pattern='.dds.res')) {
    cell <- str_split_fixed(string = file, pattern = '_', n = 2)[1]
    contrast <- str_split_fixed(string = file, pattern = '_', n = 2)[2]
    if (contrast %in% c('Age.dds.res', 'Age.scaled.dds.res','BMI.dds.res', 'BMI.scaled.dds.res',
                       'Lobular.inflammation.dds.res', 'MASH.CRN.score..X.8.dds.res', 
                       'MASH_vs_Control.pseudocount.dds.res','MASL_vs_Control.pseudocount.dds.res',
                        'MetALD_vs_Control.pseudocount.dds.res')) {
        next
    }
    if (cell == 'Bulk') {
        next
    }
    print(paste0(cell, ":", contrast))
    
    res <- read.table(paste0(deseq.outdir, file))
    res <- distinct(res)
    
    if (nrow(filter(res, pvalue < 0.05 & log2FoldChange < 0)) > 0) {
        write.table(str_split(unique(rownames(filter(res, pvalue < 0.05 & log2FoldChange < 0))), "[-:]", simplify=T),
                paste0(bed.dir, 'ATAC.', file, '.nominal.down.bed'), sep='\t', col.names=F, row.names=F, quote=F)
    }
    if (nrow(filter(res, pvalue < 0.05 & log2FoldChange > 0)) > 0) {
        write.table(str_split(unique(rownames(filter(res, pvalue < 0.05 & log2FoldChange > 0))), "[-:]", simplify=T),
                paste0(bed.dir, 'ATAC.', file, '.nominal.up.bed'), sep='\t', col.names=F, row.names=F, quote=F)
    }
    
    if (nrow(filter(res, padj < 0.05 & log2FoldChange < 0)) > 0) {
        write.table(str_split(unique(rownames(filter(res, padj < 0.05 & log2FoldChange < 0))), "[-:]", simplify=T),
                paste0(bed.dir, 'ATAC.', file, '.padj.down.bed'), sep='\t', col.names=F, row.names=F, quote=F)
    }
    if (nrow(filter(res, padj < 0.05 & log2FoldChange > 0)) > 0) {
        write.table(str_split(unique(rownames(filter(res, padj < 0.05 & log2FoldChange > 0))), "[-:]", simplify=T),
                paste0(bed.dir, 'ATAC.', file, '.padj.up.bed'), sep='\t', col.names=F, row.names=F, quote=F)
    }
}
```

```r
for (file in list.files(deseq.outdir, pattern='.dds.res')) {
    cell <- str_split_fixed(string = file, pattern = '_', n = 2)[1]
    contrast <- str_split_fixed(string = file, pattern = '_', n = 2)[2]
    if (contrast %in% c('Age.dds.res', 'Age.scaled.dds.res','BMI.dds.res', 'BMI.scaled.dds.res',
                       'Lobular.inflammation.dds.res', 'MASH.CRN.score..X.8.dds.res', 
                       'MASH_vs_Control.pseudocount.dds.res','MASL_vs_Control.pseudocount.dds.res',
                        'MetALD_vs_Control.pseudocount.dds.res')) {
        next
    }
    if (cell == 'Bulk') {
        next
    }
    print(paste0(cell, ":", contrast))
    
    res <- read.table(paste0(deseq.outdir, file))
    res <- distinct(res)
    
    if (nrow(filter(res, pvalue < 0.05 & log2FoldChange < 0)) > 0) {
        write.table(str_split(unique(rownames(filter(res, pvalue < 0.05 & log2FoldChange < 0))), "[-:]", simplify=T),
                paste0(bed.dir, 'H3K27me3_ATAC_Peaks.', file, '.nominal.down.bed'), sep='\t', col.names=F, row.names=F, quote=F)
    }
    if (nrow(filter(res, pvalue < 0.05 & log2FoldChange > 0)) > 0) {
        write.table(str_split(unique(rownames(filter(res, pvalue < 0.05 & log2FoldChange > 0))), "[-:]", simplify=T),
                paste0(bed.dir, 'H3K27me3_ATAC_Peaks.', file, '.nominal.up.bed'), sep='\t', col.names=F, row.names=F, quote=F)
    }
    
    if (nrow(filter(res, padj < 0.05 & log2FoldChange < 0)) > 0) {
        write.table(str_split(unique(rownames(filter(res, padj < 0.05 & log2FoldChange < 0))), "[-:]", simplify=T),
                paste0(bed.dir, 'H3K27me3_ATAC_Peaks.', file, '.padj.down.bed'), sep='\t', col.names=F, row.names=F, quote=F)
    }
    if (nrow(filter(res, padj < 0.05 & log2FoldChange > 0)) > 0) {
        write.table(str_split(unique(rownames(filter(res, padj < 0.05 & log2FoldChange > 0))), "[-:]", simplify=T),
                paste0(bed.dir, 'H3K27me3_ATAC_Peaks.', file, '.padj.up.bed'), sep='\t', col.names=F, row.names=F, quote=F)
    }
}
```
