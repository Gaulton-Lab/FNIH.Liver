```r
#looping through cell types by making ^ into a function
get_per_sample_gex_SUMS <- function(cell.type, filename){
    print(paste(cell.type,Sys.time()))

    #pull out rows of gex.counts where BC Ident matches cell.type
    if (cell.type=='Bulk') {
        bcs <- names(Idents(adata))
    } else {
        bcs <- names(Idents(adata)[Idents(adata) == cell.type])
    }
    counts <- gex.counts[,colnames(gex.counts) %in% bcs]
    print(dim(counts))

    #initialize the matrix of sample gex
    counts.df <- as.data.frame(rep(0,length(row.names(gex.counts))))
    row.names(counts.df) <- row.names(gex.counts)
    colnames(counts.df) <- c('temp')

    #go through samples and calculate sum of gex values
    for (sample in samples){
        sample_cols <- colnames(counts) %in% sample_bcs[[sample]]
        counts.cut <- counts[,sample_cols]
        
        #if only one bc, this becomes a vector which is an issue
        if (typeof(counts.cut) == 'double'){
            mean.counts <- counts.cut
        #if there are NO bcs, this will return NA (just return 0 for everything)
        } else if(length(colnames(counts.cut)) == 0){
            mean.counts <- rep(0,length(row.names(counts)))
        } else {
            mean.counts <- rowSums(counts.cut)
        }
        counts.df <- cbind(counts.df,as.data.frame(mean.counts))
     }
    fin.counts.df <- counts.df[,-c(1)]
    colnames(fin.counts.df) <- samples
    
    #export df
    write.table(fin.counts.df, filename, sep='\t',quote=FALSE)
}
```

```r
##### NAME YOUR FILES #####
for (cell.type in c(unique_cell_types, 'Bulk')){
    filename <- paste(outdir, cell.type, '_perdonor.RNA.counts.tsv', sep = "")
    get_per_sample_gex_SUMS(cell.type, filename)
}
```

# TPM - RNA

```r
#looping through cell types by making ^ into a function
calculate_TPM <- function(cell.type, tpm.filename, pseudobulk.filename, gene.info){
    x <- read.table(paste(pseudobulk.filename))
    
    x$length <- gene.info[rownames(x),]$length
    x <- dplyr::select(x, length, everything())
    
    #calculate rpk (as a df)
    rpk <- (x[,2:ncol(x)]/x[,1])
    #remove length col from x 
    x2 <- x[,-1]
    
    tpm <- rpk
    for (i in 1:ncol(rpk)){
        tpm[,i] <- rpk[,i]/(sum(rpk[,i])/1e6)
    }
    
    #export df
    write.table(tpm, tpm.filename, sep='\t',quote=FALSE)
}
```

```r
#looping through cell types by making ^ into a function
get_per_sample_gex_SUMS_ATAC_2_assay <- function(cell.type, filename){
    print(paste(cell.type,Sys.time()))

    #pull out rows of gex.counts.1 where BC Ident matches cell.type
    if (cell.type=='Bulk') {
        bcs <- names(Idents(adata))
    } else {
        bcs <- names(Idents(adata)[Idents(adata) == cell.type])
    }
    counts.1 <- gex.counts.1[,colnames(gex.counts.1) %in% bcs]
    print(dim(counts.1))
    counts.2 <- gex.counts.2[,colnames(gex.counts.2) %in% bcs]
    print(dim(counts.2))

    #initialize the matrix of sample gex
    counts.df <- as.data.frame(rep(0,(nrow(gex.counts.1) + nrow(gex.counts.2))))
    row.names(counts.df) <- c(row.names(gex.counts.1), row.names(gex.counts.2))
    colnames(counts.df) <- c('temp')

    #go through samples and calculate sum of gex values
    for (sample in samples){
        sample_cols <- colnames(counts.1) %in% sample_bcs[[sample]]
        counts.1.cut <- counts.1[,sample_cols]
        counts.2.cut <- counts.2[,sample_cols]
        
        #if only one bc, this becomes a vector which is an issue
        if (typeof(counts.1.cut) == 'double'){
            mean.counts.1 <- counts.1.cut
        #if there are NO bcs, this will return NA (just return 0 for everything)
        } else if(length(colnames(counts.1.cut)) == 0){
            mean.counts.1 <- rep(0,length(row.names(counts.1.cut)))
        } else {
            mean.counts.1 <- rowSums(counts.1.cut)
        }
            
        #if only one bc, this becomes a vector which is an issue
        if (typeof(counts.2.cut) == 'double'){
            mean.counts.2 <- counts.2.cut
        #if there are NO bcs, this will return NA (just return 0 for everything)
        } else if(length(colnames(counts.2.cut)) == 0){
            mean.counts.2 <- rep(0,length(row.names(counts.2.cut)))
        } else {
            mean.counts.2 <- rowSums(counts.2.cut)
        }
        mean.counts <- c(mean.counts.1,mean.counts.2)
        counts.df <- cbind(counts.df,as.data.frame(mean.counts))
     }
    fin.counts.df <- counts.df[,-c(1)]
    colnames(fin.counts.df) <- samples
    
    #export df
    write.table(fin.counts.df, filename, sep='\t',quote=FALSE)
}
```

```r
##### NAME YOUR FILES #####
for (cell.type in c(unique_cell_types, 'Bulk')){
    filename <- paste(outdir, cell.type, '_perdonor.ATAC.counts.tsv', sep = "")
    get_per_sample_gex_SUMS_ATAC_2_assay(cell.type, filename)
}
```

# Calculate CPM for ATAC

CPM = 1000000*(peak counts/depth)


```r
#looping through cell types by making ^ into a function
calculate_CPM <- function(cell.type, cpm.filename, pseudobulk.filename){
    x <- read.table(paste(pseudobulk.filename))
    
    cpm <- x
    for (i in 1:ncol(x)){
        cpm[,i] <- 1e6 * (x[,i]/sum(x[,i]))
    }
    
    #export df
    write.table(cpm, cpm.filename, sep='\t',quote=FALSE)
}
```
