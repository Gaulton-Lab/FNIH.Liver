```python

for x in range(len(library_ids)):
    library_id = library_ids[x]
    print(library_id)
    
    donor_id = donor_ids[x]
    print(donor_id)
    
    # read the raw cell counts data before scanpy normalization (not necessarily whole number integers as destriping of the 
    # 2 micron bins was done
    # prior to segmentation)
    adata_raw = sc.read_h5ad(f"{b2c_path}{library_id}/{library_id}_cdata_b2c.h5ad")
    adata_final = sc.read_h5ad(f"{tg_path}/{donor_id}/adata_0.5_tangram_filt.h5ad")
    subset_cells = adata_final.obs_names 
    subset_genes = adata_final.var_names.str.upper() #convert temporarily to uppercase to properly subset raw anndata
    adata_subset = adata_raw[subset_cells]
    adata_subset.var_names = adata_subset.var_names.str.lower() #temporarily lower
    
    genes_subset = set(adata_subset.var_names)
    genes_final = set(adata_final.var_names)

    unique_to_large = genes_subset - genes_final
    
    print(f"{len(unique_to_large)} genes unique to the larger object:")
    print(list(unique_to_large)[:20])  #
    
    adata_subset_filt = adata_subset[:, ~adata_subset.var_names.isin(unique_to_large)].copy()
    
    print("checking order")
    print("Checking cell order")
    print(adata_subset_filt.obs_names.equals(adata_final.obs_names))

    print("Checking gene order")
    print(adata_subset_filt.var_names.equals(adata_final.var_names))
    
    print("reordering")
    adata_subset_filt = adata_subset_filt[:, adata_final.var_names].copy()
    
    print("Checking gene order")
    print(adata_subset_filt.var_names.equals(adata_final.var_names))
    
    
    adata_subset_filt.obs['pred_cell_types'] = adata_final.obs['pred_cell_types']
    
    sc.pp.calculate_qc_metrics(adata_subset_filt, percent_top=(10, 20, 50, 150), inplace=True)


    print(sc.pl.violin(
    adata_subset_filt,
    keys="total_counts",
    groupby="pred_cell_types",
    rotation=90,
    stripplot=False,  # optionally show points
    jitter=0.4,
    size=1
    ))
    
    adata_li.append(adata_subset_filt)
```
