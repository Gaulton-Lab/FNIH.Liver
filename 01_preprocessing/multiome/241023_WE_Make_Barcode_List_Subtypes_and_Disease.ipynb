```r
library(dplyr)
library(tidyr)
```

```r
meta <- read.table('/nfs/lab/projects/nash_nafld_liver/downstream_outputs/chromvar_SC/seurat_meta.txt', 
                  sep='\t', header=T)
dim(meta)
head(meta)
```

```r
meta$ct_disease <- paste0(meta$celltype, "_", meta$condition)

dim(meta)
head(meta)
```

```r
cell <- unique(meta$cellsubtype)

filter(meta, cell==cellsubtype)
```

```r
write.table(unique(meta$cellsubtype), 
           paste0('/nfs/lab/tscc/welison/FNIH.Liver/peak-call-pipeline/subtypes.list.txt'),
           sep='\t', quote=F, row.names=F, col.names=F)
```

```r
for (cell in unique(meta$cellsubtype)) {
    write.table(row.names(filter(meta, cell==cellsubtype)), 
               paste0('/nfs/lab/tscc/welison/FNIH.Liver/peak-call-pipeline/barcodes/cellsubtype/', cell, '.barcode.prefix.txt'),
               sep='\t', quote=F, row.names=F, col.names=F)
}
```

```r
cell <- unique(meta$ct_disease)

filter(meta, cell==ct_disease)
```

```r
write.table(sort(unique(meta$ct_disease)), 
           paste0('/nfs/lab/tscc/welison/FNIH.Liver/peak-call-pipeline/celltypes_by_disease.list.txt'),
           sep='\t', quote=F, row.names=F, col.names=F)
```

```r
for (cell in unique(meta$ct_disease)) {
    write.table(row.names(filter(meta, cell==ct_disease)), 
               paste0('/nfs/lab/tscc/welison/FNIH.Liver/peak-call-pipeline/barcodes/celltypes_by_disease/', cell, '.barcode.prefix.txt'),
               sep='\t', quote=F, row.names=F, col.names=F)
}
```
