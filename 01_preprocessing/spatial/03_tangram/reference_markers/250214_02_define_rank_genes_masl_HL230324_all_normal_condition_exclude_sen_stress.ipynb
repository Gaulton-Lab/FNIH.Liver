```python
sc.tl.rank_genes_groups(adata_masl_heps, groupby="cellsubtype", use_raw=False)
```

```python
import pandas as pd
import csv
```

```python
markers_df_adata_masl_heps = pd.DataFrame(adata_masl_heps.uns["rank_genes_groups"]["names"]).iloc[:, :]
```

```python
markers_df_adata_masl_heps
```

```python
# Iterate through each column and save it to a text file
for celltype in markers_df_adata_masl_heps.columns:
    #celltype = 'Hepatocytes.Senescent'
        # Extract the genes for the current cell type
    genes = markers_df_adata_masl_heps[celltype].dropna()  # Drop any NaN values if present
    
    # Add quotes around each gene
    genes_with_quotes = [f'{gene}' for gene in genes]
    #print(genes.head())
    # Create a dataframe with column name "x"
    output_df = pd.DataFrame(genes_with_quotes, columns=["x"])
    
    
    print(output_df.head())
    # Save to a text file named after the cell type
    output_df.to_csv(f"/tscc/projects/ps-epigen/users/cmiciano/Liver/RNA/scripts/zone_gene_lists/MASL_alldonors_nosen_stress_{celltype}_RNA_Multiome_RankGenesALL.txt",
                     index=False, sep="\t",quoting=csv.QUOTE_ALL)
```
