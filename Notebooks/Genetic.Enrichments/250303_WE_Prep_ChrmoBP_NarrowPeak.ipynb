```r
library(dplyr)
library(stringr)
library(tidyr)
```

```r
MVP.cred.set.bed <- read.table('/nfs/lab/tscc/welison/FNIH.Liver/GWAS/NAFLD.TRANS.MVP.2021.credset.hg38.tsv',
                              sep='\t', header=T)

dim(MVP.cred.set.bed)
head(MVP.cred.set.bed)
```

```r
# SCHEMA: ["chr", "start", "end", "1" - name, "2" - score, "3" - strand, "4" - signal value, "5" - pValue, "6" - qValue, "summit"- ]

MVP.narrow.1k <- mutate(MVP.cred.set.bed, start=start.hg38-500, end=start.hg38+499, summit=500,
      strand='+',name=paste0(chr.hg38,':',end.hg38,':',NEA,':',EA),
      chr=paste0('chr',chr.hg38)) %>%
    select(chr, start, end, 
       name, score=PP, strand,
       signal=Beta, pValue=P, qValue=SE, summit)

dim(MVP.narrow.1k)
head(MVP.narrow.1k)
```

```r
write.table(MVP.narrow.1k, '/nfs/lab/tscc/welison/FNIH.Liver/13_chromBPNet_Peaks/variant_prediction/chrombpnet_plotting/250303_WE_MVP_hg38_credible_set_1k.narrowPeak',
           sep='\t', col.names=F, row.names=F, quote=F)
```
